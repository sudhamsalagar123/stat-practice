```python
df=pd.read_csv("Covid_m.csv")# loaded the data set
```

```python
df.head(3)
```

```python
df["age"].mean() # we calculated the mean of age by function
```

```python
mn=np.mean(df["age"])  # stroing the mean in mn variable
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(5,3))
sns.histplot(x="age",data=df,bins=[i for i in range(10,90,10)])
plt.plot([mn for i in range(0,90)],[i for i in range(0,90)],c="red",label="mean")
plt.legend()
plt.show()
```

```python
df.isnull().sum()
```

```python
md=np.median(df["age"])
```

```python
sns.histplot(x="age",data=df,bins=[i for i in range(0,81,10)])
plt.plot([md for i in range(0,100)],[i for i in range(0,100)],c="yellow",label="median")
plt.legend()
plt.show()
```
