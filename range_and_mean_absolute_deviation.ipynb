we are practicing the mean absolute deviation


* the mean absolute deviation is the average distance between each data points and the mean it gives the idea about the variability in data set




```python
sec_a=np.array([75,65,73,68,72,67])
sec_b=np.array([90,47,43,96,93,51])
no=np.array([1,2,3,4,5,6])
```

```python
mean_a=np.mean(sec_a)
mean_a
```

```python
mean_b=np.mean(sec_b)
mean_b
```

```python
mad_a=sec_a-mean_a
mad_a
mad_b=sec_b-mean_b
mad_a,mad_b
```

```python
plt.figure(figsize=(10,3))
plt.scatter(sec_a,no,label="sec_a")
plt.scatter(sec_b,no,label="sec_b",color="red")
plt.plot([70,70,70,70,70,70] ,no,color="blue",label="mean")
plt.legend()
plt.show()
```

```python
mad_aa=np.sum(np.abs(sec_a-mean_a)/len(sec_a))
```

```python
mad_bb=np.sum(np.abs(sec_b-mean_b)/len(sec_b))
```
